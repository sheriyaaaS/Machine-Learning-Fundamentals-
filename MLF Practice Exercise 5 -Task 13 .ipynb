```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
gender_count = df["Sex"].value_counts()
gender_count
```

```python
plt.figure(figsize=(6, 4))
gender_count.plot(kind="bar")
plt.title("Distribution of Patients by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.xticks(rotation=0)
plt.show()
```

```python
drug_count = df["Drug"].value_counts()
print(drug_count)
print(drug_count.idxmax(), "=", drug_count.max())
```

```python
drug_proportion = df["Drug"].value_counts(normalize=True) * 100
print(drug_proportion)
```

```python
plt.figure(figsize=(7, 4))
drug_proportion.plot(kind="bar")
plt.title("Proportion of Each Drug")
plt.xlabel("Drug")
plt.ylabel("Percentage")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(df["Age"], bins=10, edgecolor="black")
plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(6, 6))
plt.pie(gender_count,labels=gender_count.index,autopct="%1.1f%%")
plt.title("Male and Female Patients")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="BP", hue="Drug")
plt.title("Drug Prescription Across Different BP Levels")
plt.xlabel("Blood Pressure Level")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Drug", y="Age")
plt.title("Age Distribution Across Different Drug Types")
plt.xlabel("Drug Type")
plt.ylabel("Age")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Age", y="Na_to_K")
plt.title("Relationship Between Age and Na_to_K Ratio")
plt.xlabel("Age")
plt.ylabel("Na_to_K Ratio")
plt.show()
```
